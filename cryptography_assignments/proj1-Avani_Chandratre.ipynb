```python
from enigma.rotors.rotor import Rotor
from enigma.plugboard import Plugboard
from enigma.machine import EnigmaMachine
import string

# Inputs
plaintext = 'ATTACKXATXXXXXXATXATLANTICXZXISLAND'
ciphertext = 'WVUVJCSQBFLEJGFNIZNIGYGOCWSUVNCIIIA'
abc = string.ascii_uppercase

# Set up the Enigma machine
def create_enigma_machine():
    r1 = Rotor('rotor1', 'EKMFLGDQVZNTOWYHXUSPAIBRCJ', ring_setting=0, stepping='Q')
    r2 = Rotor('rotor2', 'BDFHJLCPRTXVZNYEIWGAKMUSQO', ring_setting=5, stepping='V')
    r3 = Rotor('rotor3', 'ESOVPZJAYQUIRHXLNFTGKDCMWB', ring_setting=10, stepping='J')
    reflector = Rotor('Reflector B', 'YRUHQSLDPXNGOKMIEBFZCWVJAT')
    pb = Plugboard.from_key_sheet('AK BZ CG DL FU HJ MX NR OY PW')
    return EnigmaMachine([r1, r2, r3], reflector, pb)

# Check all positions (AAA-ZZZ)
for a in abc:
    for b in abc:
        for c in abc:
            decryption_setting = a + b + c
            machine = create_enigma_machine()
            machine.set_display(decryption_setting)
            encrypted = machine.process_text(plaintext)

            if encrypted == ciphertext:
                print("Initial Position:", decryption_setting)
                exit()
```
