```python
import os
```

```python
def Alice_program():
    # Prompt for user input
    plaintext = input("Enter the message to encrypt: ")
    plaintext_bytes = plaintext.encode('utf-8')

    # Generate random key of the same length as plaintext
    key = os.urandom(len(plaintext_bytes))

    # Encrypt using XOR
    ciphertext = bytes([p ^ k for p, k in zip(plaintext_bytes, key)])

    # Convert to hex and save to files
    ciphertext_hex = ciphertext.hex()
    key_hex = key.hex()

    print("\nCiphertext (hex):", ciphertext_hex)

    # Step 5: Save to files
    with open("ciphertext.txt", "w") as cfile:
        cfile.write(ciphertext_hex)

    with open("key.txt", "w") as kfile:
        kfile.write(key_hex) 
```

```python
def Bob_program():
    # Read the ciphertext and key from files
    with open("ciphertext.txt", "r") as cipher_file:
        ciphertext_hex = cipher_file.read()
    with open("key.txt", "r") as key_file:
        key_hex = key_file.read()

    # Convert from hex to bytes
    ciphertext = bytes.fromhex(ciphertext_hex.strip())
    key = bytes.fromhex(key_hex.strip()) 

    # Check if key and ciphertext lengths match
    if len(ciphertext) != len(key):
        print("Error: Key and ciphertext lengths do not match!")
        return

    # Step 4: Decrypt using XOR
    decrypted_bytes = bytes([c ^ k for c, k in zip(ciphertext, key)])

    # Step 5: Convert to string and display
    plaintext = decrypted_bytes.decode('utf-8')
    print("Decrypted message:", plaintext)
```

```python
Alice_program()
Bob_program()
```
